```python
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import numpy as np
```

```python
x = np.linspace(0, 10, 100)
c = np.cos(x)
s = np.sin(x)
y = np.tan(x)
z = np.exp(x)
```

### **API Orientada a Objetos com Matriz**

```python
# Utilizando o metodo subplot podemos criar varios gráficos dentro de uma mesma figura.

# fig é a figura que contém os subplots, e axs é uma matriz de eixos (subplots).

fig, axs = plt.subplots(2, 2, figsize=(10, 8))  # 2 linhas, 2 colunas

# Primeiro subplot
axs[0,0].plot(x, c, label='Cosseno')
axs[0,0].set_title('Onda Cossenoidal')
axs[0,0].legend()

# Segundo subplot
axs[0,1].plot(x, s, label='Seno', color='r')
axs[0,1].set_title('Onda Senoidal')
axs[0,1].legend()

# Terceiro subplot
axs[1,0].plot(x, y, label='Tangente', color='g')
axs[1,0].set_title('Onda Tangencial')
axs[1,0].legend()
axs[1, 0].set_ylim(-10, 10)  # Limitar o eixo y para melhor visualização

# Quarto subplot
axs[1,1].plot(x, z, label='Exponencial', color='y')
axs[1,1].set_title('Função Exponencial')
axs[1,1].legend()
axs[1, 1].set_ylim(0, 10000)  # Limitar o eixo y para melhor visualização

# Ajustar o layout: plt.tight_layout() ajusta automaticamente o espaçamento entre os subplots para que eles não se sobreponham.
plt.tight_layout()

# Exibir a figura
plt.show()
```

### **Subplots Descompactados**

```python
fig, ((ax1, ax4), (ax3, ax2)) = plt.subplots(2, 2, figsize=(10, 8))

ax1.plot(x, c, label='Cosseno')
ax1.set_title('Onda Cossenoidal')
ax1.legend()

ax2.plot(x, s, label='Seno', color='r')
ax2.set_title('Onda Senoidal')
ax2.legend()

ax3.plot(x, y, label='Tangente', color='g')
ax3.set_title('Onda Tangencial')
ax3.legend()
ax3.set_ylim(-10, 10)

ax4.plot(x, z, label='Exponencial', color='y')
ax4.set_title('Função Exponencial')
ax4.legend()
ax4.set_ylim(0, 10000)

plt.tight_layout
plt.show()
```

### **Pyplot**

```python
plt.figure(figsize=(10, 8))

# 1º Subplot (Canto superior esquerdo)
plt.subplot(2, 2, 1)
plt.plot(x, c, label='Cosseno')
plt.title('Onda Cossenoidal')
plt.legend()

# 2º Subplot (Canto superior direito -> correspondente ao seu ax4)
plt.subplot(2, 2, 2)
plt.plot(x, z, label='Exponencial', color='y')
plt.title('Função Exponencial')
plt.legend()
plt.ylim(0, 10000)

# 3º Subplot (Canto inferior esquerdo -> correspondente ao seu ax3)
plt.subplot(2, 2, 3)
plt.plot(x, y, label='Tangente', color='g')
plt.title('Onda Tangencial')
plt.legend()
plt.ylim(-10, 10)

# 4º Subplot (Canto inferior direito -> correspondente ao seu ax2)
plt.subplot(2, 2, 4)
plt.plot(x, s, label='Seno', color='r')
plt.title('Onda Senoidal')
plt.legend()

plt.tight_layout()  # Nota: precisa dos parênteses () para executar!
plt.show()
```

### **GridSpec**

```python
# 1. Cria a figura principal
fig = plt.figure(figsize=(10, 8))

# 2. Define a grade de 3 linhas por 2 colunas
gs = gridspec.GridSpec(3, 2, figure=fig)

# --- Subplot 1: Canto Superior Esquerdo ---
ax1 = fig.add_subplot(gs[2, :])
ax1.plot(x, c, label='Cosseno')
ax1.set_title('Onda Cossenoidal')
ax1.legend()

# --- Subplot 2: Meio Esquerdo ---
ax3 = fig.add_subplot(gs[1, 0])
ax3.plot(x, y, label='Tangente', color='g')
ax3.set_title('Onda Tangencial')
ax3.legend()
ax3.set_ylim(-10, 10)

# --- Subplot 3: Coluna da Direita (Linhas 0 e 1) ---
ax2 = fig.add_subplot(gs[:2, 1])
ax2.plot(x, s, label='Seno', color='r')
ax2.set_title('Onda Senoidal (Destaque Principal)')
ax2.legend()

# --- Subplot 4: Base da Figura (Linha 2, Toda a largura) ---
ax4 = fig.add_subplot(gs[0,0])
ax4.plot(x, z, label="Exponencial", color="y")
ax4.set_title("Função Exponencial")
ax4.legend()
ax4.set_ylim(0, 500)

plt.tight_layout()
plt.show()
```
